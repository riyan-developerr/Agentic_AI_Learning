# Session 3 Summary: GitHub, README, and LinkedIn

## Key Learnings

1. **Git** saves snapshots of your project on your computer. **GitHub** stores them online. A **commit** is one snapshot with a message. A **push** uploads your commits to GitHub.
2. **`.gitignore`** lists files Git must skip: `.env`, `venv/`, and `__pycache__/`.
3. **`requirements.txt`** lets anyone install your libraries. **`.env.example`** shows which settings are needed without exposing your real key.
4. A **README** answers three questions:
   - What is this?
   - How do I run it?
   - What does it look like? (an example, plus what you learned)
5. Your **GitHub and LinkedIn profiles** are set up. Recruiters look at them before your resume.

## Important Points to Keep in Mind

- **Your repo is your proof.** A clean, public project with a clear README counts for more than a list of courses.
- **Commit small and often,** with clear messages like "Add router with Pydantic validation".
- **Check `git status` before every `git add .`** to make sure nothing secret is included.
- **Write the README for a stranger** who has never seen your project.
- **Post about one specific thing you learned** each week. Short and honest posts build your name over time.
- **Pin only your best repos.** Quality beats quantity.

## Mistakes to Avoid

- **Committing `.env` or any API key.** If it happens, delete the key in the provider's dashboard and create a new one right away. Deleting the file from GitHub later is not enough, because the history keeps it.
- **Adding `.gitignore` after the first commit.** It cannot remove files that Git already tracks.
- **Vague commit messages** like "update" or "final".
- **A repo with no README,** or one that only has the project name.
- **Leaving unfinished or messy repos pinned or featured.** Hide them until they are ready.
- **Copying code you don't understand.** In interviews you will be asked to explain every part.
- **Waiting for perfect before sharing.** Ship a small version, then improve it.
- **Going quiet for weeks.** Consistency (one commit or one post a week) beats occasional bursts.

## Self-Test Answers

1. **Commit vs push:** a commit saves a snapshot on your computer, and a push uploads it to GitHub.
2. **`.env.example`:** it shows others how to set up `.env` without exposing the real API key.
3. **README:** it should tell a visitor what the project is, how to run it, and what it looks like.

## One-Line Takeaway

> Your work only counts if others can find it, run it, and understand it, so keep it public, clean, and safe.
